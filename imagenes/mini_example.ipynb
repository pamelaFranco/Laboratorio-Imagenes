```python
# Instalar nilearn sin alterar las librerías base de Colab
!pip install -q nilearn

import nilearn.datasets as datasets
from nilearn import plotting

# 1. Cargar la plantilla MNI152 (devuelve un objeto Nifti1Image)
mni_template = datasets.load_mni152_template(resolution=1)

# 2. Mostrar la información del objeto (usando .get_filename() o .shape)
print(f"Dimensiones del volumen (matriz): {mni_template.shape}")
print(f"Ruta temporal del archivo: {mni_template.get_filename()}")

# 3. Visualizar los tres cortes ortogonales centrados en el origen (0, 0, 0)
plotting.plot_anat(
    mni_template,
    title="Plantilla MNI152 (T1w 1mm)",
    display_mode='ortho',
    cut_coords=(0, 0, 0)
)
```
